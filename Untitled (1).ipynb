```python
%pip install pandas seaborn matplotlib
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 解决中文乱码
plt.rcParams["font.family"] = ["SimHei", "WenQuanYi Micro Hei", "Heiti TC"]
plt.rcParams['axes.unicode_minus'] = False
```

```python
# 直接写数据集
data_text = """姓名,语文,数学,英语
小明,82,95,88
小红,76,80,90
小张,90,88,85
小李,65,72,78
小王,88,92,84
小赵,55,60,70
小周,78,84,81
小吴,94,96,92
小郑,70,68,75
小钱,85,83,86
"""
from io import StringIO
df = pd.read_csv(StringIO(data_text))
df.head()
```

```python
# 查看信息、缺失值
print("数据基本信息：")
print(df.info())
print("\n缺失值统计：")
print(df.isnull().sum())
df_clean = df.dropna()
```

```python
# 描述统计
print("描述性统计：")
print(df_clean.describe())
```

```python
# 绘制成绩直方图
df_clean[["语文","数学","英语"]].hist(figsize=(9,5))
plt.suptitle("各科成绩分布直方图")
plt.show()
```

```python
# 相关性热力图
corr = df_clean[["语文","数学","英语"]].corr()
sns.heatmap(corr, annot=True, cmap="Blues")
plt.title("科目相关性热力图")
plt.show()
```
